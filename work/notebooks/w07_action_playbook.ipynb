## 1. Ranked actions + reason codes

### Ranked Actions + Reason Codes

The action queue turns the validated model output into practical content actions.

Actions are ranked using the observed model output and the available evidence. The ranking is intended for decision support, not automatic execution.

Each recommendation has a reason code so a human reviewer can understand why the action was prioritized.

The main reason-code categories are:

* **HIGH_PRIORITY** — stronger observed evidence supports taking the action.
* **REFRESH** — the content appears to need updating based on the observed decay/refresh pattern.

* **MAINTAIN** — the available evidence does not indicate a strong need for change.
* **REVIEW** — the result is uncertain and should receive additional human review before action.

The queue is directional: it helps decide what to review first rather than claiming that the model has proven what action will definitely succeed.
